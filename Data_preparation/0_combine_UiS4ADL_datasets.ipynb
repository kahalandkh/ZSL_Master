Expanding the UiS4ADL dataset used in Fadi et al. (2024) from 30 to 43 subjects.

```python
import os
import warnings
import pandas as pd
import numpy as np

# Suppress warnings from Python and external libraries
os.environ["PYTHONWARNINGS"] = "ignore"  # Suppress Python warnings
warnings.filterwarnings("ignore")
```

```python
fs = 100 # sampling frequency in Hz
```

```python
# Loading the csv file with the old data
old_data = pd.read_csv("../Data/UiS4ADL/Raw/Old/UiS4ADL-100hz.csv")
print(old_data.head())

# Loading the csv file with the new data
new_data = pd.read_csv("../Data/UiS4ADL/Raw/New/all_data.csv")
print(new_data.head())

timestep = 10
```

```python
# Adding a new timestamp column to the new data
for fileID in sorted(new_data['fileID'].unique()):
    mask = new_data['fileID'] == fileID
    new_data.loc[mask, 'timestamp'] = np.arange(mask.sum()) * timestep
new_data
```

```python
def concat_csv_with_metadata_and_files(df1_old, df2_new):
    try:
        # Keep only the columns in df2_new that are also in df1_old
        common_columns = [col for col in df2_new.columns if col in df1_old.columns]
        df2_new = df2_new[common_columns]

        # Concatenate the two DataFrames vertically
        combined_df = pd.concat([df1_old, df2_new], ignore_index=True)
        return combined_df

    except Exception as e:
        print(f"Error processing files: {e}")

df = concat_csv_with_metadata_and_files(old_data, new_data)
df
```

```python
num_subjects = df['subject'].nunique()
print(f"Number of subjects after expansion: {num_subjects}")
```

## Save the combined dataset

```python
 # Save the combined DataFrame to a new CSV file
output_file = f"../Data/UiS4ADL/Raw/UiS4ADL_{fs}hz_raw.csv"

# Remove the file if it already exists
if os.path.exists(output_file):
    os.remove(output_file)
    print(f"{output_file} has been removed successfully.")
else:
    print(f"{output_file} does not exist.")
    
# Save the combined DataFrame to a CSV file
df.to_csv(output_file, index=False)
print(f"The dataset was saved to {output_file}")
```
