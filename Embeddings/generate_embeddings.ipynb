Generate, save, and load embeddings for different models and activity description variants.

```python
import sys
sys.path.append('..')

import numpy as np

from Embeddings.Models.embedding_generator import EmbeddingGenerator
from Embeddings.Semantics.descriptions import DESCRIPTION_TYPES, get_descriptions
```

```python
DESCRIPTION_TYPES.keys()
```

```python
print("Available description types:")
for desc_type in DESCRIPTION_TYPES:
    descriptions = get_descriptions(desc_type)
    print(f"\n{desc_type.upper()}:")
    print(f"Example: {list(descriptions.values())[0]}")
```

### Generate embeddings

```python
generator = EmbeddingGenerator(output_dir='../Data/Embeddings/')
```

```python
results = {}

for model in generator.MODELS:
    for desc_type, descriptions in DESCRIPTION_TYPES.items():
        print(f"\nModel: {model}")
        print(f"Description type: {desc_type}")
        
        descriptions = get_descriptions(desc_type)
        embeddings, filepath = generator.generate_and_save(
            descriptions, model, desc_type, use_prompt=True)

        if model.startswith("clip"):
            key = f"{model.replace('/', '_').replace('-', '_').lower()}_{desc_type}"
        else:
            key = f"{model.split('/')[-1]}_{desc_type}"
        results[key] = embeddings
```
